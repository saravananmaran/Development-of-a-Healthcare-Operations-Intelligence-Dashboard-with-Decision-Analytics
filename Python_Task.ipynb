**Most commonly asked Python interview questions**

```python
s = "Python"
print(s[::-1])
```

```python
s = input("Enter a string: ")

if s == s[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
n = int(input("Enter a number: "))
fact = 1

for i in range(1, n + 1):
    fact *= i

print("Factorial =", fact)
```

```python
n = int(input("Enter number of terms: "))

a, b = 0, 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
n = int(input("Enter a number: "))

if n < 2:
    print("Not Prime")
else:
    prime = True

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            prime = False
            break

    if prime:
        print("Prime")
    else:
        print("Not Prime")
```

```python
numbers = [10, 45, 23, 67, 12]

largest = numbers[0]

for n in numbers:
    if n > largest:
        largest = n

print("Largest =", largest)
```

```python
s = input("Enter a string: ")

count = 0

for ch in s.lower():
    if ch in "aeiou":
        count += 1

print("Vowels =", count)
```

```python
a = 10
b = 20

a, b = b, a

print(a)
print(b)
```

```python
n = int(input("Enter a number: "))

temp = n
digits = len(str(n))
total = 0

while temp > 0:
    digit = temp % 10
    total += digit ** digits
    temp //= 10

if total == n:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

```python
n = int(input("Enter a number: "))

total = 0

while n > 0:
    total += n % 10
    n //= 10

print("Sum =", total)
```

```python
numbers = [1, 2, 2, 3, 3, 3, 4]

frequency = {}

for n in numbers:
    frequency[n] = frequency.get(n, 0) + 1

print(frequency)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even = [n for n in numbers if n % 2 == 0]

print(even)
```
